```python
churn_rate = df.groupby('Customer service calls')['Churn'].mean()

churn_rate 
```

```python
churn_rate = df.groupby('Customer service calls')['Churn'].mean()

fig, ax = plt.subplots(figsize=(8, 4))
churn_rate.plot(kind='bar', ax=ax)
ax.set_title('Churn rate by number of customer service calls')
ax.set_xlabel('Customer service calls')
ax.set_ylabel('Churn rate')
plt.show()
```

```python
Q1 = df['Total day minutes'].quantile(0.25)
Q3 = df['Total day minutes'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['Total day minutes'] < lower) | (df['Total day minutes'] > upper)]
len(outliers)
```

```python
z = (df['Total day minutes'] - df['Total day minutes'].mean()) / df['Total day minutes'].std()
len(df[z.abs() > 3])
```

```python
import seaborn as sns

corr = df.select_dtypes('number').corr()

fig, ax = plt.subplots(figsize=(10, 8))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm',
            vmin=-1, vmax=1, ax=ax)
plt.show()
```
