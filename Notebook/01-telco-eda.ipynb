```python
df.groupby('Churn').agg({
    'tenure': 'mean',
    'MonthlyCharges': 'mean'
})
```

```python
df.groupby('Churn').agg(
    avg_tenure=('tenure', 'mean'),
    avg_monthly=('MonthlyCharges', 'mean')
)
```

```python
df.groupby('Churn').agg(
    avg_tenure=('tenure', 'mean'),
    median_tenure=('tenure', 'median'),
    n_customers=('customerID', 'count')
)
```
